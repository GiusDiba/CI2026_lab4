### Importing all the problems


```python
from search_algorithm import search
from utils.klotski_problems import *

def solve_problem(prb):
    print(f"Solving problem: {prb}")
    sol = search(prb.getBoard())
    if sol["solved"]:
        for mv in sol["moves"]:
            prb.board.movePiece(mv[0], mv[1])
        assert prb.isSolved()
        print(f"Problem {prb._id} solved in {len(sol["moves"])} moves after visiting {sol["visited_states"]} states")
        print(f"Moves:\n{sol["moves"]}")
    else:
        print(f"Problem {prb._id} could not be solved. Visited states: {sol["visited_states"]}")
```
